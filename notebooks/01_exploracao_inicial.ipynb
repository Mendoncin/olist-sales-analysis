```python
tabelas = {}

for arquivo in arquivos:
    nome = arquivo.stem.replace("olist_", "").replace("_dataset", "")
    tabelas[nome] = pd.read_csv(arquivo)

for nome, df in tabelas.items():
    print(f"\n{nome}")
    print(f"Linhas: {df.shape[0]} | Colunas: {df.shape[1]}")
    print("Colunas:", df.columns.tolist())
```

```python
chaves = {
    "orders": "order_id",
    "customers": "customer_id",
    "products": "product_id",
    "sellers": "seller_id",
    "order_reviews": "review_id",
}

for tabela, coluna in chaves.items():
    df = tabelas[tabela]

    print(f"\n{tabela}")
    print(f"Total de linhas: {len(df)}")
    print(f"Valores únicos: {df[coluna].nunique()}")
    print(f"Valores duplicados: {df[coluna].duplicated().sum()}")
    print(f"Valores nulos: {df[coluna].isna().sum()}")
```

```python
clientes = tabelas["customers"]

# Quantos identificadores únicos de compradores existem?
print(clientes["customer_unique_id"].nunique())

# Quantos registros de clientes existem?
print(clientes["customer_id"].nunique())

# Quantos compradores aparecem em mais de um registro?
frequencia = clientes["customer_unique_id"].value_counts()
print((frequencia > 1).sum())

# Exemplos de compradores recorrentes
print(frequencia.head(10))
```

```python
itens = tabelas["order_items"]

print("Total de registros:", len(itens))

print(
    "order_id duplicados:",
    itens["order_id"].duplicated().sum()
)

print(
    "Chave composta duplicada:",
    itens.duplicated(
        subset=["order_id", "order_item_id"]
    ).sum()
)

print(
    "Nulos nas chaves:",
    itens[["order_id", "order_item_id"]].isna().sum()
)
```

```python
pedidos_ids = set(tabelas["orders"]["order_id"])
itens_ids = set(tabelas["order_items"]["order_id"])

print("Pedidos sem itens:", len(pedidos_ids - itens_ids))
print("Itens com pedidos inexistentes:", len(itens_ids - pedidos_ids))
```

```python
pedidos_sem_itens = pedidos[
    pedidos["order_id"].isin(itens["order_id"])
]

pedidos_sem_itens["order_status"].value_counts()
```

```python
pagamentos = tabelas["order_payments"]

print(
    "Pedidos distintos:",
    pagamentos["order_id"].nunique()
)

print(
    "Chave composta duplicada:",
    pagamentos.duplicated(
        subset=["order_id", "payment_sequential"]
    ).sum()
)

print(
    "Pagamentos sem pedido existente:",
    (~pagamentos["order_id"].isin(pedidos["order_id"])).sum()
)

print(
    "Registros com chaves nulas:",
    pagamentos[["order_id", "payment_sequential"]].isna().sum()
)
```

```python
pedidos_sem_pagamento = pedidos[
    ~pedidos["order_id"].isin(pagamentos["order_id"])
]

pedidos_sem_pagamento
```

```python
avaliacoes = tabelas["order_reviews"]

print("Total de avaliações:", len(avaliacoes))

print(
    "Pedidos distintos:",
    avaliacoes["order_id"].nunique()
)

print(
    "review_id duplicados:",
    avaliacoes["review_id"].duplicated().sum()
)

print(
    "order_id duplicados:",
    avaliacoes["order_id"].duplicated().sum()
)

print(
    "Chave composta duplicada:",
    avaliacoes.duplicated(
        subset=["review_id", "order_id"]
    ).sum()
)

print(
    "Linhas inteiramente duplicadas:",
    avaliacoes.duplicated().sum()
)
```

```python
# Avaliações cujo review_id aparece mais de uma vez
reviews_repetidas = avaliacoes[
    avaliacoes["review_id"].duplicated(keep=False)
]

display(
    reviews_repetidas[
        ["review_id", "order_id", "review_score"]
    ].sort_values("review_id").head(12)
)
```

```python
# Pedidos que aparecem em múltiplas avaliações
pedidos_multiplas_reviews = avaliacoes[
    avaliacoes["order_id"].duplicated(keep=False)
]

display(
    pedidos_multiplas_reviews[
        ["order_id", "review_id", "review_score"]
    ].sort_values("order_id").head(12)
)
```

```python
produtos = tabelas["products"]
vendedores = tabelas["sellers"]
itens = tabelas["order_items"]

print("Produtos distintos:", len(produtos))
print("Produtos vendidos:", itens["product_id"].nunique())

print(
    "Itens com produtos inexistentes:",
    (~itens["product_id"].isin(produtos["product_id"])).sum()
)

print("\nVendedores distintos:", len(vendedores))
print("Vendedores com vendas:", itens["seller_id"].nunique())

print(
    "Itens com vendedores inexistentes:",
    (~itens["seller_id"].isin(vendedores["seller_id"])).sum()
)
```

```python
geo = tabelas["geolocation"]
categorias = tabelas["product_category_name_translation"]

print("=== GEOLOCALIZAÇÃO ===")

print("Total de registros:", len(geo))

print(
    "Prefixos de CEP distintos:",
    geo["geolocation_zip_code_prefix"].nunique()
)

print(
    "Prefixos duplicados:",
    geo["geolocation_zip_code_prefix"].duplicated().sum()
)

print("\n=== CATEGORIAS ===")

print("Total de traduções:", len(categorias))

print(
    "Categorias distintas:",
    categorias["product_category_name"].nunique()
)

print(
    "Categorias duplicadas:",
    categorias["product_category_name"].duplicated().sum()
)

print(
    "Produtos sem categoria:",
    produtos["product_category_name"].isna().sum()
)
```

```python
# Categorias utilizadas pelos produtos
categorias_produtos = set(
    produtos["product_category_name"].dropna()
)

# Categorias existentes na tabela de tradução
categorias_traduzidas = set(
    categorias["product_category_name"].dropna()
)

print("Categorias utilizadas:", len(categorias_produtos))
print("Categorias traduzidas:", len(categorias_traduzidas))

print(
    "Categorias sem tradução:",
    categorias_produtos - categorias_traduzidas
)

print(
    "Categorias traduzidas não utilizadas:",
    categorias_traduzidas - categorias_produtos
)

print(
    "Chaves nulas na tradução:",
    categorias["product_category_name"].isna().sum()
)
```

```python
geo = tabelas["geolocation"]
clientes = tabelas["customers"]
vendedores = tabelas["sellers"]

ceps_geo = set(geo["geolocation_zip_code_prefix"])
ceps_clientes = set(clientes["customer_zip_code_prefix"])
ceps_vendedores = set(vendedores["seller_zip_code_prefix"])

print("=== CLIENTES ===")
print("CEPs distintos:", len(ceps_clientes))
print("CEPs sem correspondência:", len(ceps_clientes - ceps_geo))

print("\n=== VENDEDORES ===")
print("CEPs distintos:", len(ceps_vendedores))
print("CEPs sem correspondência:", len(ceps_vendedores - ceps_geo))
```
